```python
#A1,A2
#there were many indentation errors in the chunk. Nearly all unindented lines required an indent. To keep code clean, I will not indicate every single needed indent
import random
random.seed(42)
#Given code to generate inventory
def generate_inventory(num_categories=5, num_products_per_category=10):
    categories = ['Hand soap', 'Moisturizer', 'Sunscreen', 'Body oil', 'Shampoo'] #switched to proper quotation marks, remove 'and' shampoo
    inventory = {}
    for category in categories:
        inventory[category] = []
        for _ in range(num_products_per_category):
            product = {
                'Quantity': random.randint(10, 100),
                'Unit Price': round(random.uniform(5.0, 500.0), 2),
                'Expiration Date': random.randint(-2, 365),
                'Advanced Ordered': random.randint(0, 8)
            }
      
            inventory[category].append(product)
           
    return inventory #switched inventories to inventory
inventory = generate_inventory() # added this per outside help to view code
```

```python
#A3iii Min and Max expiration days appearing anywhere in the inventory
exp_days = []                               

for category in inventory:        #used technique learned in class to start and append empty list, needed to do this as it is a dictionary, not dataframe
    for product in inventory[category]:      
        exp_days.append(product['Expiration Date'])

print("Minimum expiration days:", min(exp_days))
print("Maximum expiration days:", max(exp_days))
```

```python
#B1
qualifying = []  #empty list append technique

for category in inventory:
    for product in inventory[category]:
        if (product['Unit Price'] > 200
                and product['Expiration Date'] < 30
                and product['Quantity'] > product['Advanced Ordered']):

            record = {             #need to basically rewrite inventory in order to make proper ammend
                'Category': category,
                'Quantity': product['Quantity'],
                'Unit Price': product['Unit Price'],
                'Expiration Date': product['Expiration Date'],
                'Advanced Ordered': product['Advanced Ordered'],
                'Inventory Value': product['Quantity'] * product['Unit Price']
            }

            qualifying.append(record)
print(qualifying)
```

```python
#B2 print the information for all qualifying products
for p in qualifying:
    print(p['Category'], "Qty:", p['Quantity'],
          "Unit Price:", p['Unit Price'],
          "Exp:", p['Expiration Date'],
          "Adv Ordered:", p['Advanced Ordered'],
          "Inventory Value:", round(p['Inventory Value'], 2))
```

```python
# B3: Total number of qualifying products
print("Qualifying products:", len(qualifying))
```

```python
# B4: Aggregate inventory value
aggregate = 0         #0 technique similar to empty list technique but numeric
for p in qualifying:     
    aggregate = aggregate + p['Inventory Value']

print("Aggregate value:", round(aggregate, 2))
```

```python
#B5: Calculate avg quantity / avg advanced ordered
avg_qty = sum(p['Quantity'] for p in qualifying) / len(qualifying) #used AI to learn a comprehension shortcut
avg_adv = sum(p['Advanced Ordered'] for p in qualifying) / len(qualifying)

if avg_adv == 0:
    print("Average Advanced Ordered is zero/undefined")
else:
    print("Avg Quantity / Avg Advanced Ordered:", round(avg_qty / avg_adv, 2))
```

```python
# B6: Category contributing the largest dollar amount
top_category = None
top_category_value = 0 #used 0 technique 

for category in inventory:
    total = 0
    for p in qualifying:
        if p['Category'] == category:
            total = total + p['Inventory Value']

    print(category, round(total, 2))

    if total > top_category_value: #used AI to assist with this part after entering pseudo code
        top_category = category
        top_category_value = total

print("Top contributing category:", top_category, round(top_category_value, 2))
```

```python
#C1 Normalize unit price against the highest
highest_unit_price = 0 #used the 0 technique but replace it if number is larger
for p in qualifying:
    if p['Unit Price'] > highest_unit_price:
        highest_unit_price = p['Unit Price']

for p in qualifying:
    p['Normalized Unit Price'] = round(p['Unit Price'] / highest_unit_price, 4)

print(highest_unit_price)
```

```python
#C2 Highest, lowest, and the lowest normalized value

lowest_unit_price = qualifying[0]['Unit Price'] #used similar syntax but cant use 0 for a min
for p in qualifying:
    if p['Unit Price'] < lowest_unit_price:
        lowest_unit_price = p['Unit Price']

print("Most expensive qualifying product:", highest_unit_price)
print("Least expensive qualifying product:", lowest_unit_price)
print("Least expensive qualifying product:", round(lowest_unit_price / highest_unit_price, 4))
```

```python
#C3 Sorted from highest normalized price to lowest
sorted_products = sorted(qualifying, key=lambda p: p['Normalized Unit Price'], reverse=True) #AI was used and helped me learn about sort and lambda

for p in sorted_products:
    print(p['Category'],
          "Unit Price:", p['Unit Price'],
          "Normalized:", p['Normalized Unit Price'],
          "Qty:", p['Quantity'],
          "Days left:", p['Expiration Date'])
```

A normalized price close to 1 means the product costs almost as much as the most expensive item in the group. Those products lose the company the most money per unit if they expire, so they should be sold or discounted first. Numbers closer to 0 are the cheaper products, where expiring costs much less.

```python
#D Does removing the most expensive qualifying product change the top category

# Di Find the most expensive qualifying product
most_expensive = qualifying[0]
for p in qualifying:
    if p['Unit Price'] > most_expensive['Unit Price']:
        most_expensive = p

print("Removed:", most_expensive['Category'], most_expensive['Unit Price'])

# Build a new list with everything except that one
remaining = []
for p in qualifying:
    if p is not most_expensive: #learned about is not function
        remaining.append(p)

# Same category totals as B6, but on the smaller list
new_top_category = None
new_top_value = 0

for category in inventory:
    total = 0
    for p in remaining:
        if p['Category'] == category:
            total = total + p['Inventory Value']

    print(category, round(total, 2))

    if total > new_top_value:
        new_top_category = category
        new_top_value = total

print("Before:", top_category, round(top_category_value, 2))
print("After:", new_top_category, round(new_top_value, 2))

if new_top_category == top_category:
    print("The top category did NOT change.")
else:
    print("The top category DID change, from", top_category, "to", new_top_category)
```
