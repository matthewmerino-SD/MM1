```python
n1 = float(input("Input number 1:"))
n2 = float(input("Input number 2:"))
print("1:add, 2:subtract, 3: multiply, 4 divide")
c = int(input("Input operator indicator:"))
if c ==1 :
    print(n1+n2)
elif c==2:
    print(n1-n2)
elif c==3:
    print(n1*n2)
elif c==4:
    print(n1/n2)
else: print("You did not select a valid operator indicator")
    

```

```python
n1 = float(input("Input number 1:"))
n2 = float(input("Input number 2:"))
print("Select your operator - 1:+, 2:-, 3:*, 4:/")
c = input("Input operator indicator:")


if c =="+" :
    print(n1+n2)
elif c=="-":
    print(n1-n2)
elif c=="*":
    print(n1*n2)
elif c=="/":
    if n2 == 0:
        print("Zero error")
    else: print(n1/n2)
else: print("You did not select a valid operator indicator")
```
